# Classification of MNIST dataset


```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score


import autograd.numpy as np
import pandas as pd

import torch
from torch.utils.data import DataLoader, TensorDataset
from torch import nn


# Custom imports
from source.mnist_preprocessing import X_train, X_test, y_train, y_test, ITERATIONS, NP_RANDOM_SEED, TORCH_SEED, HIDDEN_LAYERS, BATCH_SIZE, ETA_VALUES, LAMBDA_VALUES, MOMENTUM
from source.plotting import plot_heatmap
from source.utils import pytorch_loop, pytorch_model_mnist_fn
```

## Classification of MNIST dataset with PyTorch

```python
# convert to Pytorch tensors
y_train_idx = np.asarray(y_train, dtype=np.int64)  
y_test_idx  = np.asarray(y_test,  dtype=np.int64)


X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train_idx, dtype=torch.long)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test_idx, dtype=torch.long)

# Create dataset
batch_value = 256
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_dl = DataLoader(train_dataset, batch_value, shuffle=True)
```

```python

EPOCHS = 100
BATCH_SIZE = 256
cost_f = 'CrossEntropy'
weight_decay_val = 0 # no lambda
rho1_val = 0.9
rho2_val = 0.999
n_class = 10

VERBOSE = True
IMG_DATA = True # ensure flatten in model generation for image classification
SAVE_FIGURE = True
SHOW_PLOT = True
```

```python
# Combinations of hidden layers to test
list_hidden_layers = [
    #(8,),
    #(16,),
    #(50,),
    #(8, 16),
    (50, 100),
    #(16, 32, 16),
    #(50, 100, 50, 100, 50)
]

# activation function to test
list_activation_functions = [nn.ReLU(), nn.LeakyReLU(), nn.Softmax(dim=1)]

penalties = ['L1', 'L2']
```

```python
# Loop through list of hidden layers and activation functions
for hidden in list_hidden_layers:
    for act_f in list_activation_functions:
        for pen in penalties:

            print(f'Hidden layers: {hidden, act_f}, Activation function: {act_f}')
            print()

            
            model_adam = lambda hidden=hidden, act_f=act_f: pytorch_model_mnist_fn(
                hidden_units_list=hidden,
                input_data=X_train_tensor,
                num_classes=n_class,
                pytorch_activation=act_f,
                image_data=IMG_DATA,
                verbose=VERBOSE
            )


            adam_mnist = pytorch_loop(model_adam, ETA_VALUES, LAMBDA_VALUES, 'ADAM', EPOCHS,
                            X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor, cost_func=cost_f,
                            rho=rho1_val, rho2=rho2_val, weight_decay=weight_decay_val, batch_size=BATCH_SIZE, verbose=True)
            
            
            # Save heatmaps
            if SAVE_FIGURE:

                task = 'part_f'
                algo = 'pytorch'
                plot_type = 'heatmap'

                filename_values = f'{task}-{plot_type}-{hidden}-activation{act_f}-penalty{pen}-iter{EPOCHS}-batch{batch_value}-{algo}'


                plot_heatmap(adam_mnist, 'Accuracy', y_axis_scientific=True, show_plot=SHOW_PLOT, title=f'Heatmap with Accuracy values for ADAM - ', filename = 'adam_' + filename_values)
            
```

## Classification of MNIST dataset with LogisticRegression from Scikit-learn

```python
# Initialize the model
model = LogisticRegression(solver='saga', max_iter=ITERATIONS, random_state=NP_RANDOM_SEED)
# Train the model
model.fit(X_train, y_train)
# Make predictions on the test set
y_pred = model.predict(X_test)
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")

```
