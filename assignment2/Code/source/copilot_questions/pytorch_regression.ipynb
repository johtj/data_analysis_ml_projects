# Question to Copilot for regression with Pytorch

Vis eksempelkode for polynomial regresjon med PyTorch

Code modified for use in part_c

```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# Generer syntetiske data
torch.manual_seed(0)
x = torch.linspace(-3, 3, 100).unsqueeze(1)
y = 0.5 * x**3 - x**2 + x + torch.randn_like(x) * 3  # Cubic polynomial + støy

# Lag polynomfunksjoner ved å utvide input med polynomtermer
def make_poly_features(x, degree):
    return torch.cat([x**i for i in range(1, degree + 1)], dim=1)

degree = 3
x_poly = make_poly_features(x, degree)

# Definer en enkel lineær modell
model = nn.Linear(degree, 1)

# Tapfunksjon og optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Tren modellen
epochs = 1000
for epoch in range(epochs):
    y_pred = model(x_poly)
    loss = criterion(y_pred, y)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Visualiser resultatet
with torch.no_grad():
    y_fit = model(x_poly)
    plt.scatter(x.numpy(), y.numpy(), label='Data')
    plt.plot(x.numpy(), y_fit.numpy(), color='red', label='Polynomial Fit')
    plt.legend()
    plt.title('Polynomial Regression with PyTorch')
    plt.show()
```
